```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches
import pandas as pd

colors = [
    '#1f77b4',  # Muted Blue
    '#ff7f0e',  # Safety Orange
    '#2ca02c',  # Cooked Asparagus Green
    '#d62728',  # Brick Red
    '#9467bd',  # Muted Purple
    '#8c564b',  # Chestnut Brown
    '#e377c2',  # Raspberry Pink
    '#7f7f7f',  # Middle Gray
    '#bcbd22',  # Curry Yellow-Green
    '#17becf'   # Blue-Muted Cyan
]
```

```python
dat300K_1 = np.genfromtxt('R&TBiCMOS_IHP/gummel_filter1_300k.txt')
dat77K_2 = np.genfromtxt('R&TBiCMOS_IHP/gummel_filter2_77k.txt')
dat300K_2  = np.genfromtxt('R&TBiCMOS_IHP/gummel_300k.txt')
# dat_77K_new = np.genfromtxt('R&TBiCMOS_IHP/newpcb/T2_CT.txt')
```

```python
# --- Corrected rcParams Block ---
plt.rcParams.update({
    ## Text and Font Settings
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "axes.labelsize": 25,
    "font.size": 16,
    "legend.fontsize": 16,
    "text.latex.preamble": r"\usepackage{amsmath}",

    ## Tick Settings
    "xtick.labelsize": 22,
    "ytick.labelsize": 22,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 8,
    "ytick.major.size": 8,
    "xtick.minor.size": 4,
    "ytick.minor.size": 4,
    
    ## Axes Line Width (for a bolder border)
    "axes.linewidth": 1.5,
    
    ## Turn the default grid OFF in rcParams
    "axes.grid": False,
})
```

```python
# --- Plotting Code ---
fig, ax1 = plt.subplots(figsize=(10, 7)) # Slightly larger figure for better readability
ax2 = ax1.twinx()

ax1.semilogy(dat300K_1[1:-1,0], dat300K_1[1:-1,1]*1e3, color = 'red', lw=3, label = r'$I_B$')
ax1.semilogy(dat300K_1[1:-1,0], dat300K_1[1:-1,2]*1e3, color = 'blue', lw=3, label = r'$I_C$')
ax2.semilogy(dat300K_1[1:-1,0], dat300K_1[1:-1,3], color ='green',linestyle = 'dashed', lw=3, label = r'$\beta$')
# --- Axis Labels and Titles ---
ax1.set_xlabel("$V_{BE} = V_{CE}$ [V]", fontsize=16) # Added units
ax1.set_ylabel("Current [mA]", fontsize=16, color='black') # Explicitly set color for clarity

ax2.set_ylabel("Current Gain ($\\beta$)", fontsize=16, color='green') # Added LaTeX for beta, set color
ax2.tick_params(axis='y', labelcolor='green') # Make ax2 tick labels green to match the line

# --- Grid ---
ax1.grid(True, which="both", ls="-", color='0.7', linewidth=0.5) # Both major and minor grid lines, light grey
# You could also add a grid specifically to ax2 if you want it to be different
# ax2.grid(True, which="both", ls=":", color='0.8', linewidth=0.3)

# --- Ticks and Appearance ---
ax1.tick_params(axis='both', which='major', labelsize=12)
ax1.tick_params(axis='both', which='minor', labelsize=10) # Optional: if you want minor tick labels
ax2.tick_params(axis='both', which='major', labelsize=12)
ax2.tick_params(axis='both', which='minor', labelsize=10) # Optional

# Add minor ticks, which are very useful for log plots
ax1.minorticks_on()
ax2.minorticks_on()


# --- Legends ---
# Combine legends from both axes into one
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='lower right', fontsize=12, frameon=True, fancybox=True, shadow=True) # Unified legend, placed for clarity

# --- Title ---
fig.suptitle("BJT Gummel Plot: Current vs. Voltage and Current Gain @300K", fontsize=18, y=1.02) # More descriptive title, slightly higher

# --- Layout and Saving ---
fig.tight_layout(rect=[0, 0.03, 1, 0.98]) # Adjust layout to prevent title overlap
plt.show()
```

```python
# New way to read the file, test 
import os
def process_csv_files_in_folder(folder_path):
    """
    Reads all CSV files in a folder, extracts data after the line
    containing "Dimension2", and saves it to a new file with the same name
    in a 'processed' subdirectory.

    Args:
        folder_path (str): The path to the folder containing the CSV files.
    """
    # Create an output directory if it doesn't exist
    output_folder = os.path.join(folder_path, "processed")
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # List all files in the given folder
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            
            # Find the line number to start reading from
            start_row = 0
            with open(file_path, 'r') as f:
                for i, line in enumerate(f):
                    if "Dimension2" in line:
                        start_row = i + 1
                        break
            
            # Read the CSV from the determined start row
            if start_row > 0:
                try:
                    df = pd.read_csv(file_path, skiprows=start_row)
                    
                    # Save the processed dataframe to a new CSV file
                    output_file_path = os.path.join(output_folder, filename)
                    df.to_csv(output_file_path, index=False)
                    print(f"Processed '{filename}' and saved to '{output_file_path}'")
                except Exception as e:
                    print(f"Could not process file {filename}. Error: {e}")
            else:
                print(f"Keyword 'Dimension2' not found in {filename}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_segmented_curves(df, x_col, y_col, num_curves, ax=None,
                          curves_to_plot=None, title=None, xlabel=None,
                          ylabel=None, **plot_kwargs):
    """
    Plots selected or all curves from concatenated sweeps.

    Args:
        df (pd.DataFrame): DataFrame with the plotting data.
        x_col (str): Column name for the x-axis.
        y_col (str): Column name for the y-axis.
        num_curves (int): The TOTAL number of distinct curves available in the data.
        ax (matplotlib.axes.Axes, optional): The axes object to plot on.
        curves_to_plot (list of int, optional): A list of specific curve indices
            to plot (e.g., [0, 2, 5]). If None, all curves are plotted.
        title (str, optional): Title for the plot.
        xlabel (str, optional): Label for the x-axis.
        ylabel (str, optional): Label for the y-axis.
        **plot_kwargs: Additional keyword arguments for plt.plot().
    """
    # --- 1. Validation ---
    total_points = len(df)
    if total_points % num_curves != 0:
        print("Error: Data cannot be split evenly.")
        return None, None
    points_per_curve = total_points // num_curves

    # --- 2. Determine which curves to iterate over ---
    if curves_to_plot is None:
        # If no specific curves are requested, plot all of them
        plot_indices = range(num_curves)
    else:
        # If a list is provided, validate it and use it
        if any(i >= num_curves or i < 0 for i in curves_to_plot):
            print(f"Error: `curves_to_plot` has an invalid index. Use indices from 0 to {num_curves - 1}.")
            return None, None
        plot_indices = curves_to_plot

    # --- 3. Setup for plotting ---
    if ax is None:
        fig, ax = plt.subplots(figsize=(12, 8))
    else:
        fig = ax.get_figure()

    label = plot_kwargs.pop('label', None)
    is_first_plot = True  # Use a flag to apply label only to the first curve plotted

    # --- 4. Plotting Loop ---
    for i in plot_indices:  # Loop over only the selected indices
        start_index = i * points_per_curve
        end_index = start_index + points_per_curve
        curve_data = df.iloc[start_index:end_index]

        # Apply label only to the first curve in our selection
        if is_first_plot:
            ax.plot(curve_data[x_col], curve_data[y_col], label=label, **plot_kwargs)
            is_first_plot = False # The label is now used, don't use it again
        else:
            ax.plot(curve_data[x_col], curve_data[y_col], **plot_kwargs)

    # --- 5. Customization ---
    if title: ax.set_title(title)
    if xlabel: ax.set_xlabel(xlabel)
    if ylabel: ax.set_ylabel(ylabel)
    ax.grid(True)

    return fig, ax
```

```python
IcVce_RT['Ic uA'] = IcVce_RT[' IC']*1e6 
IcVce_CT1['Ic uA'] = IcVce_CT1[' IC']*1e6 
# 1. Create a base plot with some other data (e.g., a load line)
fig, ax = plt.subplots(figsize=(12, 8))
plot_segmented_curves(
    df=IcVce_RT,
    x_col=' VC',
    y_col='Ic uA',
    num_curves=12,
    ax=ax,  # <-- This is the magic! We pass the axes to the function.
    color='red',
    linestyle = 'dashed',
    curves_to_plot=[0],
    lw = 2,
    label=r'$I_B = 1 \mu A ~@300K$' # label will be used by ax.legend()
)

plot_segmented_curves(
    df=IcVce_RT,
    x_col=' VC',
    y_col='Ic uA',
    num_curves=12,
    ax=ax,  # <-- This is the magic! We pass the axes to the function.
    color='red',
    lw = 2,
    curves_to_plot=[1],
    label=r'$I_B = 2 \mu A ~@300K$' # label will be used by ax.legend()
)

# 2. Now, call our function and tell it to draw on the *existing* axes
plot_segmented_curves(
    df=IcVce_CT1,
    x_col=' VC',
    y_col='Ic uA',
    num_curves=8,
    ax=ax,  # <-- This is the magic! We pass the axes to the function.
    color='blue',
    lw = 2,
    linestyle = 'dashed',
    curves_to_plot=[3],
    label= r'$I_B = 1 \mu A ~@77K$' # label will be used by ax.legend()
)

# 2. Now, call our function and tell it to draw on the *existing* axes
plot_segmented_curves(
    df=IcVce_CT1,
    x_col=' VC',
    y_col='Ic uA',
    num_curves=8,
    lw = 2,
    ax=ax,  # <-- This is the magic! We pass the axes to the function.
    color='blue',
    curves_to_plot=[6],
    label= r'$I_B = 2 \mu A ~ @77K$' # label will be used by ax.legend()
)

# 3. Add final touches to the combined plot
# 3. Add final touches to the combined plot
# ax.set_title(r"$I_C - V_{CE}$ Sweep", fontsize = 20)
ax.set_xlabel(r"$V_{CE} ~ [V]$", fontsize = 25)
ax.set_ylabel(r" $I_C ~[\mu A]$", fontsize = 25)
# Or, more commonly, set both at the same time:
ax.tick_params(axis='both', which='major', labelsize=18)
ax.set_ylim(0, 1500) # Set y-axis limits for a nice view
ax.set_xlim(0,2)
ax.legend(fontsize = 17) # Display the labels for both plots
ax.grid(True, which="both", ls="-", color='0.7', linewidth=0.5)

plt.savefig('plot/IcVce_sweep_low_Ib.png')

plt.show()
```
